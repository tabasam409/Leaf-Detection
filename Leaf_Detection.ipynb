```python
import numpy as np
import cv2
import os

# Function to calculate Euclidean distance between two points (feature vectors)
def euclidean_distance(point1, point2):
    return np.sqrt(np.sum((point1 - point2) ** 2))

# Function to find the k-nearest neighbors
def knn_classify(features, labels, query_point, k=3):
    distances = []
    
    # Calculate the distance from the query point to each data point
    for i, feature in enumerate(features):
        distance = euclidean_distance(query_point, feature)
        distances.append((distance, labels[i]))
    
    # Sort distances and get the k-nearest neighbors
    distances.sort(key=lambda x: x[0])
    nearest_neighbors = distances[:k]
    
    # Get the most common label among the k-nearest neighbors
    label_counts = {}
    for neighbor in nearest_neighbors:
        label = neighbor[1]
        label_counts[label] = label_counts.get(label, 0) + 1
    
    # Return the label with the highest count
    return max(label_counts, key=label_counts.get)

# Function to extract features from an image (using image shape for simplicity)
def extract_features(image_path):
    # Check if image exists
    if not os.path.exists(image_path):
        print(f"Image does not exist at: {image_path}")
        return None
    
    # Attempt to read the image using OpenCV
    image = cv2.imread(image_path)
    if image is None:
        print(f"Could not read image at: {image_path}")
        return None
    
    # Simple feature extraction: image shape (height, width, channels)
    features = image.shape  # Example feature: image dimensions (height, width, channels)
    return features

# Directory path to the dataset
dataset_path = r'C:\Users\Zain\Documents\dataset'  # Update with your dataset path
features = []
labels = []

# Loop through each plant folder and extract features
for plant_folder in os.listdir(dataset_path):
    plant_folder_path = os.path.join(dataset_path, plant_folder)
    
    if os.path.isdir(plant_folder_path):
        # Iterate over the images inside the current plant folder
        for image_name in os.listdir(plant_folder_path):
            image_path = os.path.join(plant_folder_path, image_name)
            
            # Extract features for each image in the dataset
            feature = extract_features(image_path)
            if feature is not None:
                features.append(feature)
                labels.append(plant_folder)  # The folder name is the label (plant species)

# Classify each image in the dataset using k-NN
for plant_folder in os.listdir(dataset_path):
    plant_folder_path = os.path.join(dataset_path, plant_folder)
    
    if os.path.isdir(plant_folder_path):
        for image_name in os.listdir(plant_folder_path):
            image_path = os.path.join(plant_folder_path, image_name)
            
            # Extract features from the current image (query image)
            query_features = extract_features(image_path)
            
            if query_features is not None:
                # Classify the image using k-NN
                predicted_label = knn_classify(features, labels, np.array(query_features), k=3)
                print(f"Image: {image_name} from {plant_folder} is classified as: {predicted_label}")
```

```python
import numpy as np
import cv2
import os

# Function to calculate color histogram
def extract_color_histogram(image_path, bins=8):
    # Read the image
    image = cv2.imread(image_path)
    if image is None:
        print(f"Could not read image at: {image_path}")
        return None
    
    # Convert the image to HSV color space
    hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    
    # Calculate the histogram for each channel (Hue, Saturation, Value)
    h_hist = cv2.calcHist([hsv_image], [0], None, [bins], [0, 256])
    s_hist = cv2.calcHist([hsv_image], [1], None, [bins], [0, 256])
    v_hist = cv2.calcHist([hsv_image], [2], None, [bins], [0, 256])
    
    # Normalize the histograms
    h_hist = cv2.normalize(h_hist, h_hist).flatten()
    s_hist = cv2.normalize(s_hist, s_hist).flatten()
    v_hist = cv2.normalize(v_hist, v_hist).flatten()
    
    # Concatenate the histograms into a single feature vector
    hist = np.concatenate([h_hist, s_hist, v_hist])
    
    # Return the color histogram
    return hist

# Function to process all images in the dataset
def process_dataset(dataset_path):
    features = []
    labels = []

    # Loop through each plant folder and extract color histograms
    for plant_folder in os.listdir(dataset_path):
        plant_folder_path = os.path.join(dataset_path, plant_folder)
        
        if os.path.isdir(plant_folder_path):
            # Loop through all images in the current plant folder
            for image_name in os.listdir(plant_folder_path):
                image_path = os.path.join(plant_folder_path, image_name)
                
                # Extract the color histogram for each image
                feature = extract_color_histogram(image_path)
                if feature is not None:
                    features.append(feature)
                    labels.append(plant_folder)  # The folder name is the label (plant species)
    
    return features, labels

# Directory path to the dataset
dataset_path = r'C:\Users\Zain\Documents\dataset'  # Update with your dataset path

# Process the dataset and extract features
features, labels = process_dataset(dataset_path)

# Optionally, print or save the features for each image in the dataset
for feature, label in zip(features, labels):
    print(f"Label: {label}, Feature: {feature[:10]}...")  # Print first 10 values for brevity
```

```python
import numpy as np
import cv2
import os

# Function to calculate Euclidean distance between two points (feature vectors)
def euclidean_distance(point1, point2):
    return np.sqrt(np.sum((point1 - point2) ** 2))

# Function to classify a new image using k-NN
def knn_classify(features, labels, query_point, k=3):
    distances = []
    
    # Calculate the distance from the query point to each data point
    for i, feature in enumerate(features):
        distance = euclidean_distance(query_point, feature)
        distances.append((distance, labels[i]))
    
    # Sort distances and get the k-nearest neighbors
    distances.sort(key=lambda x: x[0])
    nearest_neighbors = distances[:k]
    
    # Get the most common label among the k-nearest neighbors
    label_counts = {}
    for neighbor in nearest_neighbors:
        label = neighbor[1]
        label_counts[label] = label_counts.get(label, 0) + 1
    
    # Return the label with the highest count
    return max(label_counts, key=label_counts.get)

# Function to extract color histogram
def extract_color_histogram(image_path, bins=8):
    image = cv2.imread(image_path)
    if image is None:
        print(f"Could not read image at: {image_path}")
        return None
    
    # Convert the image to HSV color space
    hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    
    # Calculate the histogram for each channel (Hue, Saturation, Value)
    h_hist = cv2.calcHist([hsv_image], [0], None, [bins], [0, 256])
    s_hist = cv2.calcHist([hsv_image], [1], None, [bins], [0, 256])
    v_hist = cv2.calcHist([hsv_image], [2], None, [bins], [0, 256])
    
    # Normalize the histograms
    h_hist = cv2.normalize(h_hist, h_hist).flatten()
    s_hist = cv2.normalize(s_hist, s_hist).flatten()
    v_hist = cv2.normalize(v_hist, v_hist).flatten()
    
    # Concatenate the histograms into a single feature vector
    hist = np.concatenate([h_hist, s_hist, v_hist])
    
    # Return the color histogram
    return hist

# Function to process all images in the dataset
def process_dataset(dataset_path):
    features = []
    labels = []

    # Loop through each plant folder and extract color histograms
    for plant_folder in os.listdir(dataset_path):
        plant_folder_path = os.path.join(dataset_path, plant_folder)
        
        if os.path.isdir(plant_folder_path):
            # Loop through all images in the current plant folder
            for image_name in os.listdir(plant_folder_path):
                image_path = os.path.join(plant_folder_path, image_name)
                
                # Extract the color histogram for each image
                feature = extract_color_histogram(image_path)
                if feature is not None:
                    features.append(feature)
                    labels.append(plant_folder)  # The folder name is the label (plant species)
    
    return features, labels

# Directory path to the dataset
dataset_path = r'C:\Users\Zain\Documents\dataset'  # Update with your dataset path

# Process the dataset and extract features
features, labels = process_dataset(dataset_path)

# Test classification for each image in the dataset using k-NN
for plant_folder in os.listdir(dataset_path):
    plant_folder_path = os.path.join(dataset_path, plant_folder)
    
    if os.path.isdir(plant_folder_path):
        for image_name in os.listdir(plant_folder_path):
            image_path = os.path.join(plant_folder_path, image_name)
            
            # Extract features from the current image (query image)
            query_features = extract_color_histogram(image_path)
            
            if query_features is not None:
                # Classify the image using k-NN
                predicted_label = knn_classify(features, labels, query_features, k=3)
                print(f"Image: {image_name} from {plant_folder} is classified as: {predicted_label}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import os

# Function to extract features from an image (dummy example, adjust to your feature extraction method)
def extract_features(image_path):
    # Example: Convert image to grayscale and compute the mean color value as a simple feature
    image = cv2.imread(image_path)
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    features = np.mean(gray_image)  # Just a placeholder; you should replace with real feature extraction
    return features

# Generate confusion matrix manually
def create_confusion_matrix(true_labels, predicted_labels, unique_labels):
    n = len(unique_labels)
    label_to_index = {label: i for i, label in enumerate(unique_labels)}
    cm = np.zeros((n, n), dtype=int)
    
    for true, pred in zip(true_labels, predicted_labels):
        true_idx = label_to_index[true]
        pred_idx = label_to_index[pred]
        cm[true_idx, pred_idx] += 1
    
    return cm, unique_labels

# Example KNN implementation
def knn_classify(features, labels, test_feature, k):
    distances = np.linalg.norm(features - test_feature, axis=1)
    nearest_indices = np.argsort(distances)[:k]
    nearest_labels = [labels[i] for i in nearest_indices]
    predicted_label = max(set(nearest_labels), key=nearest_labels.count)
    return predicted_label

# Load the dataset and extract features
dataset_dir = "path_to_your_plant_leaf_images"  # Replace with the actual path
labels = []
features = []

for species_folder in os.listdir(dataset_dir):
    species_path = os.path.join(dataset_dir, species_folder)
    if os.path.isdir(species_path):
        for image_name in os.listdir(species_path):
            image_path = os.path.join(species_path, image_name)
            labels.append(species_folder)  # Add the label based on folder name
            feature = extract_features(image_path)  # Extract features from the image
            features.append(feature)

# Convert lists to numpy arrays
features = np.array(features)
labels = np.array(labels)

# Predict labels for all images in the dataset using KNN
predicted_labels = []
for feature in features:
    predicted_label = knn_classify(features, labels, feature, k=3)
    predicted_labels.append(predicted_label)

# Ensure predicted labels and true labels have the same length
assert len(labels) == len(predicted_labels), "Mismatch between true and predicted labels."

# Create confusion matrix
unique_labels = np.unique(labels)
conf_matrix, label_names = create_confusion_matrix(labels, predicted_labels, unique_labels)

# Plot confusion matrix
plt.figure(figsize=(8, 8))
plt.imshow(conf_matrix, interpolation='nearest', cmap=plt.cm.Blues)
plt.title("Confusion Matrix")
plt.colorbar()

# Add labels and annotations
plt.xticks(ticks=np.arange(len(label_names)), labels=label_names, rotation=45)
plt.yticks(ticks=np.arange(len(label_names)), labels=label_names)
for i in range(len(label_names)):
    for j in range(len(label_names)):
        plt.text(j, i, str(conf_matrix[i, j]), ha='center', va='center', color="red")

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.tight_layout()
plt.show()
```
